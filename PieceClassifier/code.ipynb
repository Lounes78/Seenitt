```python
import torch
import matplotlib.pyplot as plt
from ChessDataset import CLASS_NAMES

mean = torch.tensor((0.485, 0.456, 0.406)).view(1,3,1,1)
std  = torch.tensor((0.229, 0.224, 0.225)).view(1,3,1,1)

def show_misclassified_fast(pieceClassifier, max_show=12):
    pieceClassifier.model_instance.eval()
    mis = []

    with torch.inference_mode():
        for batch in pieceClassifier.test_dataloader:
            if len(batch) == 2:
                img, y = batch
                extra = None
            else:
                img, y, extra = batch

            img_dev = img.to(pieceClassifier.device, dtype=torch.float32)
            y_dev = y.to(pieceClassifier.device, dtype=torch.long)

            pred = pieceClassifier.model_instance(img_dev).argmax(1)

            wrong = (pred != y_dev).nonzero(as_tuple=False).squeeze(1).tolist()
            for i in wrong:
                mis.append((img[i].cpu(), int(y[i]), int(pred[i].cpu()),
                            None if extra is None else extra[i]))
                if len(mis) >= max_show:
                    break

            # liberar refs GPU del batch
            del img_dev, y_dev, pred
            if pieceClassifier.device != "cpu":
                torch.cuda.empty_cache()

            if len(mis) >= max_show:
                break

    print(f"Misclassified (shown): {len(mis)}")
    if not mis:
        return

    cols = min(4, len(mis))
    rows = (len(mis) + cols - 1) // cols
    plt.figure(figsize=(4*cols, 4*rows))

    for k, (img_chw, yt, yp, extra_k) in enumerate(mis, 1):
        x = img_chw.unsqueeze(0)
        x = (x * std + mean).clamp(0, 1)[0].permute(1,2,0).numpy()

        ax = plt.subplot(rows, cols, k)
        ax.imshow(x)
        ax.set_title(f"true: {yt} {CLASS_NAMES[yt]}\npred: {yp} {CLASS_NAMES[yp]}", fontsize=9)
        ax.axis("off")

    plt.tight_layout()
    plt.show()
```

```python
show_misclassified_fast(pieceClassifier)
```
