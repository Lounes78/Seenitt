```python
import torch
import matplotlib.pyplot as plt
from ChessDataset import CLASS_NAMES

# denormalize (ImageNet)
mean = torch.tensor((0.485, 0.456, 0.406)).view(1,3,1,1)
std  = torch.tensor((0.229, 0.224, 0.225)).view(1,3,1,1)

def show_misclassified(pieceClassifier, max_show=12):
    pieceClassifier.model_instance.eval()

    mis = []  # (img_cpu_CHW, y, pred, extra)

    with torch.no_grad():
        for batch in pieceClassifier.test_dataloader:
            # Ajusta esto segun tu Dataset:
            # si devuelve (img, y)
            if len(batch) == 2:
                img, y = batch
                extra = None
            # si devuelve (img, y, extra) por ejemplo class_name o file_name
            else:
                img, y, extra = batch

            img = img.to(pieceClassifier.device).float()
            y = y.to(pieceClassifier.device).long()

            out = pieceClassifier.model_instance(img)
            pred = out.argmax(dim=1)

            wrong = (pred != y).nonzero(as_tuple=False).squeeze(1)
            for i in wrong.tolist():
                mis.append((
                    img[i].detach().cpu(),
                    int(y[i].detach().cpu()),
                    int(pred[i].detach().cpu()),
                    None if extra is None else extra[i]
                ))

    print(f"Misclassified: {len(mis)}")
    if len(mis) == 0:
        return

    n = min(max_show, len(mis))
    cols = min(4, n)
    rows = (n + cols - 1) // cols

    plt.figure(figsize=(4*cols, 4*rows))
    for k in range(n):
        img_chw, yt, yp, extra = mis[k]

        x = img_chw.unsqueeze(0)
        x = (x * std + mean).clamp(0, 1)[0]     # CHW in [0,1]
        x = x.permute(1,2,0).numpy()            # HWC

        ax = plt.subplot(rows, cols, k+1)
        ax.imshow(x)
        title = f"true: {yt} {CLASS_NAMES[yt]}\npred: {yp} {CLASS_NAMES[yp]}"
        if extra is not None:
            title += f"\n{extra}"
        ax.set_title(title, fontsize=9)
        ax.axis("off")

    plt.tight_layout()
    plt.show()

show_misclassified(pieceClassifier, max_show=20)
```

```python
import torch
import time
import numpy as np

@torch.no_grad()
def benchmark_inference(model, dataloader, device, warmup=10, runs=100):
    model.eval()

    # warmup
    for i, (img, *_ ) in enumerate(dataloader):
        if i >= warmup:
            break
        img = img.to(device)
        _ = model(img)

    times = []

    for i, (img, *_ ) in enumerate(dataloader):
        if i >= runs:
            break

        img = img.to(device)

        torch.cuda.synchronize()
        t0 = time.perf_counter()

        _ = model(img)

        torch.cuda.synchronize()
        t1 = time.perf_counter()

        times.append((t1 - t0) * 1000)  # ms

    times = np.array(times)
    print(f"Mean: {times.mean():.2f} ms")
    print(f"Std : {times.std():.2f} ms")
    print(f"Min : {times.min():.2f} ms")
    print(f"Max : {times.max():.2f} ms")

    return times
```

```python
benchmark_inference(
    pieceClassifier.model_instance,
    pieceClassifier.test_dataloader,
    pieceClassifier.device
)
```

# Using engine


```python
import tensorrt as trt
import pycuda.driver as cuda
import pycuda.autoinit
import numpy as np

TRT_LOGGER = trt.Logger(trt.Logger.WARNING)

def load_engine(engine_path: str) -> trt.ICudaEngine:
    with open(engine_path, "rb") as f, trt.Runtime(TRT_LOGGER) as runtime:
        engine = runtime.deserialize_cuda_engine(f.read())
    return engine

engine = load_engine("piece_resnet18_128_fp16.engine")
context = engine.create_execution_context()

# 1) Detectar nombres I/O
input_names = []
output_names = []
for i in range(engine.num_io_tensors):
    name = engine.get_tensor_name(i)
    mode = engine.get_tensor_mode(name)
    if mode == trt.TensorIOMode.INPUT:
        input_names.append(name)
    else:
        output_names.append(name)

print("inputs:", input_names)
print("outputs:", output_names)

# (Asumimos 1 input y 1 output)
inp = input_names[0]
out = output_names[0]

# 2) Setear shape (si es dinámico, igual hay que setearlo)
context.set_input_shape(inp, (1, 3, 128, 128))

# 3) Obtener shapes y alocar buffers
inp_shape = tuple(context.get_tensor_shape(inp))
out_shape = tuple(context.get_tensor_shape(out))

inp_dtype = trt.nptype(engine.get_tensor_dtype(inp))
out_dtype = trt.nptype(engine.get_tensor_dtype(out))

h_in = cuda.pagelocked_empty(int(np.prod(inp_shape)), dtype=inp_dtype)
h_out = cuda.pagelocked_empty(int(np.prod(out_shape)), dtype=out_dtype)

d_in = cuda.mem_alloc(h_in.nbytes)
d_out = cuda.mem_alloc(h_out.nbytes)

stream = cuda.Stream()

# 4) Asociar punteros al contexto
context.set_tensor_address(inp, int(d_in))
context.set_tensor_address(out, int(d_out))

def infer_trt(img_np: np.ndarray) -> np.ndarray:
    """
    img_np: (1,3,128,128) float32 (normalizada ImageNet)
    returns: logits (1,12) float32/float16 dependiendo engine
    """
    np.copyto(h_in, img_np.ravel())

    cuda.memcpy_htod_async(d_in, h_in, stream)
    context.execute_async_v3(stream_handle=stream.handle)
    cuda.memcpy_dtoh_async(h_out, d_out, stream)
    stream.synchronize()

    return np.array(h_out).reshape(out_shape)
```

```python
import time
import numpy as np
import torch

def benchmark_trt(
    infer_fn,
    img_np,
    warmup=20,
    runs=100
):
    for _ in range(warmup):
        _ = infer_fn(img_np)

    times = []
    for _ in range(runs):
        t0 = time.perf_counter()
        _ = infer_fn(img_np)
        t1 = time.perf_counter()
        times.append((t1 - t0) * 1000)  # ms

    times = np.array(times)
    print(f"Mean: {times.mean():.3f} ms")
    print(f"Std : {times.std():.3f} ms")
    print(f"Min : {times.min():.3f} ms")
    print(f"Max : {times.max():.3f} ms")

    return times
```
